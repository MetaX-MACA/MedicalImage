# 沐曦GPU运行SAM-Med3D说明文档

[SAM-Med3D](https://github.com/uni-medical/SAM-Med3D)是面向三维医学图像的通用分割模型。仅需少量三维提示点，SAM-Med3D即可精准分割多种模态下的解剖结构与病灶。其依托于大规模三维医学影像数据集 `SA-Med3D-140K`（整合了公开及授权私有数据）进行训练，具备卓越的泛化性能。

## 1. 安装
### 1.1 安装monai
> 请参考[沐曦GPU运行MONAI 1.5.2说明文档](https://github.com/MetaX-MACA/MedicalImage/blob/main/MONAI/README.md) 安装monai  
> 为简化环境部署，后续会在沐曦开发者社区提供`MONAI镜像`，请关注[沐曦开发者|镜像资源中心](https://developer.metax-tech.com/softnova/docker/)获取对应资源。

### 1.2 下载并安装SAM-Med3D

```python
%%bash
cd /opt 
git clone https://github.com/uni-medical/SAM-Med3D.git
pip install opencv-python-headless matplotlib prefetch_generator edt surface-distance medim torchio
```

### 1.3 环境验证

```python
%%bash
cd /opt/SAM-Med3D
git rev-parse HEAD
python -c "import torch; assert torch.cuda.is_available(); print('PyTorch version:', torch.__version__); print('CUDA OK:', torch.cuda.get_device_name(0))"
python -c "import monai; monai.config.print_config()"
```

## 2. 沐曦GPU 运行效果展示
运行中会自动下载[sam_med3d_turbo.pth](https://huggingface.co/blueyo0/SAM-Med3D/blob/main/sam_med3d_turbo.pth)。如已在本地，请手动修改`medim_val_single.py`文件中的`ckpt_path`.

```python
%%bash
export PYTORCH_DEFAULT_NDHWC=1
cd /opt/SAM-Med3D
sed -i.bak 's/padding_params, cropping_params = crop_transform\._compute_center_crop_or_pad(subject_canonical)/padding_params, cropping_params = crop_transform(subject_canonical)/' /opt/SAM-Med3D/utils/infer_utils.py
python medim_val_single.py
```
